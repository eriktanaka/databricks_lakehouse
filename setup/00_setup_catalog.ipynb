```python
%sql
-- project catalog
CREATE CATALOG IF NOT EXISTS olist_lakehouse
COMMENT 'Lakehouse project';

USE CATALOG olist_lakehouse;

-- schemas creation
CREATE SCHEMA IF NOT EXISTS landing
COMMENT 'Raw files';

-- Medalion layers
CREATE SCHEMA IF NOT EXISTS bronze COMMENT 'raw data, no transformation';
CREATE SCHEMA IF NOT EXISTS silver COMMENT 'clean data, typed and de-duplicated';
CREATE SCHEMA IF NOT EXISTS gold   COMMENT 'dimensional modeled, aggregated and ready for BI use';

-- volume creation
CREATE VOLUME IF NOT EXISTS landing.olist
COMMENT 'raw olist data, incrmen';
```
